```python
class Student(BaseModel):
    name:str  = Field(description="Student name")
    roll_no:int = Field(description="student roll number")
    batch:str = Field(description="student academic batch")
    cgpa:float = Field(description="student CPGA")
```

```python

def convert_marks_to_cgpa(marks:int)->float:
    """this function converts marks to cgpa"""
    return round(marks/9.5,2)
```

```python
from langchain_core.tools import Tool

def tool(func):
    return Tool.from_function(
        func=func,
        name=func.__name__,
        description=func.__doc__ or func.__name__
    )
```

```python
@tool
def convert_marks_to_cgpa(marks: int) -> float:
    """Convert marks to CGPA."""
    return round(marks / 9.5, 2)
```

```python
convert_marks_to_cgpa.__doc__
```

```python
result = convert_marks_to_cgpa.invoke({"marks": 85})
```

```python
print(result)
```

```python
from langchain_core.tools import StructuredTool

def addition_func(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

addition = StructuredTool.from_function(
    func=addition_func,
    name="addition",
    description="Add two numbers."
)

print(addition.args)
```

```python
print(addition.get_input_schema().model_json_schema())
```

```python
result = addition.invoke({"a": 10, "b": 20})
print(result)   
```

```python
from langchain_groq import ChatGroq
import os
from dotenv import load_dotenv

load_dotenv(override=True)

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0
)

print("Model initialized successfully")
```

```python
import builtins

while True:
    user_query = builtins.input("Enter your query (or type 'exit' to quit): ")

    if user_query.lower() == "exit":
        break

    ai_response = llm.invoke(user_query)
    print("AI Response:", ai_response.content)
```

```python
import builtins

message = []

while True:
    user_query = builtins.input(
        "Enter your query (or type 'exit' to quit): "
    )

    if user_query.strip().lower() == "exit":
        break

    ai_response = llm.invoke(user_query)

    if ai_response.content:
        print("AI Response:", ai_response.content)
    else:
        print("No response from AI.")
```

```python
import builtins

message = []

while True:
    user_query = builtins.input("Enter your query (or type 'exit' to quit): ")

    if user_query.strip().lower() == "exit":
        break

    message.append({"role": "user", "content": user_query})

    ai_response = llm.invoke(message)

    if ai_response.tool_calls:
        print("Tool calls requested:", ai_response.tool_calls)
    else:
        print("AI Response:", ai_response.content)

    message.append({
        "role": "assistant",
        "content": ai_response.content or ""
    })
```
